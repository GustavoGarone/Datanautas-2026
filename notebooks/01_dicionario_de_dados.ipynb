## Observações importantes para a análise

1. **Taxas:** `taxa_desvinculo`, `taxa_trancamento` e `taxa_conclusao_anual` são proporções, não percentuais. Por exemplo, `0.1331` representa 13,31%.

2. **Vínculos:** a variável `vinculos` não inclui `concluintes` em sua soma. Essa definição foi verificada nos registros do arquivo.

3. **Taxa de conclusão:** `taxa_conclusao_anual` não equivale à taxa de conclusão de uma coorte de ingressantes. Portanto, não deve ser interpretada diretamente como a probabilidade de um estudante concluir o curso.

4. **Indicadores de perfil:** as variáveis iniciadas por `mat_` representam contagens de matrículas e não categorias mutuamente exclusivas. Um estudante pode pertencer simultaneamente a vários grupos.

5. **Dados ausentes:** `co_municipio` e `municipio` não estão preenchidos para cursos EAD. A variável `sigla_ies` também apresenta valores ausentes.

6. **Análise longitudinal:** `co_curso` e `ano` formam uma chave única na tabela. A variável `presente_5_anos` permite identificar um painel balanceado de cursos com registros em todos os anos.

7. **Cursos com poucos alunos:** `poucos_alunos` utiliza o critério `vinculos < 30`, e não `matriculas < 30`. Essa distinção é importante para aplicar filtros na análise.

8. **Presença no painel:** `anos_com_alunos` contabiliza os anos em que há registros do curso, inclusive quando `matriculas = 0`. Portanto, não significa necessariamente que houve estudantes com matrícula ativa em todos esses anos.
